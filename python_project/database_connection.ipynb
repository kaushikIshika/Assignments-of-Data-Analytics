```python
def intialized_database():
    """Create the database and Books table if they do not exist."""

    # First connect without selecting a database so it can be created.

    server_config={
        key:value for key, value in db_coonection.items()
        if key!="Database"
    }

    connection=None
    cursor=None

    try:
        connection=mysql.connector.connect(server_config)
        cursor=connection.cursor()
        cursor.execute("CREATE DATABASE IF NOT EXISTS LibraryDB")
        cursor.execute("use LibraryDB")
        cursor.execute("""
        CREATE TABLE IF NOT EXISTS Books (
                book_id VARCHAR(20) PRIMARY KEY,
                title VARCHAR(150) NOT NULL,
                author VARCHAR(100) NOT NULL,
                department VARCHAR(100) NOT NULL,
                status ENUM('Available', 'Issued') NOT NULL DEFAULT 'Available',
                issued_to VARCHAR(100) DEFAULT NULL)
            """)
        connection.commit()
        return true 
    except Error as error:
        print(f"database setup: {error}")
        print("Check that MySQL Server is running and DB_CONFIG is correct.")
        return False

    finally:
        if cursor is not None:
            cursor.close()
        if connection is not None:
            connection.close()
            
            
```
